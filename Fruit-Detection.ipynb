```python
import pandas as pd
import tensorflow as tf
import os
from tensorflow.keras import layers, models, datasets
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
import numpy as np
import PIL
import PIL.Image
import pathlib
from tensorflow.keras.models import Sequential
from tensorflow import keras



train_path = "fruits/train"
test_path = "fruits/test"
validation_path = "fruits/validation"


data_dir = pathlib.Path("fruits")
train_dir = data_dir 
test_dir  = data_dir

image_count = len(list(data_dir.rglob('*.png')))
apples = list(data_dir.rglob('banana/*'))
#img =PIL.Image.open(str(apples[5]))
#display(img)

img_height = 128 
img_width = 128
batch_size = 32

train_ds = tf.keras.utils.image_dataset_from_directory(
  data_dir,
  validation_split=0.2,
  subset="training",
  seed=123,
  image_size=(img_height, img_width),
  batch_size=batch_size)

validation_ds = tf.keras.utils.image_dataset_from_directory(data_dir, validation_split =0.2, subset="validation", seed=123, 
                                                            image_size = (img_height, img_width), batch_size = batch_size)
class_names = train_ds.class_names
#print(class_names)

plt.figure(figsize=(10, 10))
#for images, labels in train_ds.take(1):
#for i in range(9):
#x = plt.subplot(3, 3, i + 1) #creates the 3 rows and 3 columns and takes the first three entries
#plt.imshow(images[i].numpy().astype("uint8")) #Converts the image tensor to a NumPy array, changes its data type to an 8-bit unsigned integer (standard for viewing images), and draws it.
   # plt.title(class_names[labels[i]]) #Looks up the text name of the label (like "cat" or "dog") and places it as a title above the image.
   # plt.axis("off")

normalization_layer = tf.keras.layers.Rescaling(1./255) #to make it easier for the neural network to work with, [0,1] instead of [0,255]
normalized_ds = train_ds.map(lambda x, y: (normalization_layer(x), y))
image_batch, labels_batch = next(iter(normalized_ds))
first_image = image_batch[0]
print(np.min(first_image), np.max(first_image))

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.cache().prefetch(buffer_size=AUTOTUNE) #cache() stores the dataset after its been loaded for the first time
validation_ds = validation_ds.cache().prefetch(buffer_size=AUTOTUNE) #prefetch prepares the data before loading ///Overall makes training faster

num_classes = len(class_names)

model = Sequential([
  layers.Rescaling(1./255, input_shape=(img_height, img_width, 3)), #shirnks the pixel values from [0,255] -> [0,1] 
  layers.Conv2D(16, 3, padding='same', activation='relu'), #first convolution layer, that detects the visual features of the images
  layers.MaxPooling2D(), #shrinks the 'image representation' 
  layers.Conv2D(32, 3, padding='same', activation='relu'), #second convolution layer with more filters/ looking at the features produced by the previous layer
  layers.MaxPooling2D(), #shrinks the image representation again
  layers.Conv2D(64, 3, padding='same', activation='relu'),#third layer with even more filters to recognize more complex features
  layers.MaxPooling2D(),
  layers.Flatten(), #turns all the data into actual numerical values to be read
  layers.Dense(128, activation='relu'), 
  layers.Dense(num_classes) #checks which class is most likely 
])


#This line of code is basically a summary of how the model is performing, Where to adjust weights and what problems to show 
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True), #A loss function to measure how accurate the predictions were
              metrics=['accuracy'])
model.summary()


epochs =10
history = model.fit(train_ds, validation_data = validation_ds, epochs = epochs) #What actually trains the data, returns the accuracy and validation accuracy
#if we begin to notice taht acc= 99% while valid_acc begins to drop. this may be a sign of overfitting

acc = history.history['accuracy']
valid_acc = history.history['val_accuracy']

loss = history.history['loss']
valid_loss = history.history['val_loss']

epochs_range = range(epochs)

plt.figure(figsize=(8, 8))
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Training Accuracy')
plt.plot(epochs_range, valid_acc, label='Validation Accuracy')
plt.legend(loc='lower right')
plt.title('Training and Validation Accuracy')

plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Training Loss')
plt.plot(epochs_range, valid_loss, label='Validation Loss')
plt.legend(loc='upper right')
plt.title('Training and Validation Loss')
plt.show()

#currently our program is suffering from overfitting, we will now use data augmentation and dropout to fix this 
data_augmentation = keras.Sequential( [layers.RandomFlip("horizontal", 
                                                         input_shape = (img_height, img_width, 3)), 
                                       layers.RandomRotation(0.1),
                                       layers.RandomZoom(0.1),])
#this right here uses randomflip, rotation, and zoom to 'skew' the image to create new images to train off of 

#outlays the images that are skewed
plt.figure(figsize=(10, 10))
for images, _ in train_ds.take(1):
  for i in range(9):
    augmented_images = data_augmentation(images)
    ax = plt.subplot(3, 3, i + 1)
    plt.imshow(augmented_images[0].numpy().astype("uint8"))
    plt.axis("off")

#Another technique to deal with overfitting is 'dropout' which basically drops out some of the networks neurons so it doesnt rely on bad habits

model = Sequential([
  layers.Rescaling(1./255), #shirnks the pixel values from [0,255] -> [0,1] 
  layers.Conv2D(16, 3, padding='same', activation='relu'), #first convolution layer, that detects the visual features of the images
  layers.MaxPooling2D(), #shrinks the 'image representation' 
  layers.Conv2D(32, 3, padding='same', activation='relu'), #second convolution layer with more filters/ looking at the features produced by the previous layer
  layers.MaxPooling2D(), #shrinks the image representation again
  layers.Conv2D(64, 3, padding='same', activation='relu'),#third layer with even more filters to recognize more complex features
  layers.MaxPooling2D(),
  layers.Dropout(0.2), #turns all the data into actual numerical values to be read
  layers.Flatten(),  
  layers.Dense(128, activation='relu'), 
  layers.Dense(num_classes, name="outputs") #checks which class is most likely 
])

model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True), #A loss function to measure how accurate the predictions were
              metrics=['accuracy'])
model.summary()

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True)
epochs =30
history = model.fit(train_ds, validation_data = validation_ds, epochs = epochs, callbacks=[early_stopping]) 

acc = history.history['accuracy']
val_acc = history.history['val_accuracy']

loss = history.history['loss']
val_loss = history.history['val_loss']

epochs_range = range(len(acc))

plt.figure(figsize=(8, 8))
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Training Accuracy')
plt.plot(epochs_range, val_acc, label='Validation Accuracy')
plt.legend(loc='lower right')
plt.title('Training and Validation Accuracy')

plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Training Loss')
plt.plot(epochs_range, val_loss, label='Validation Loss')
plt.legend(loc='upper right')
plt.title('Training and Validation Loss')
plt.show()
                            
```
